### **Завдання 1 (3 бали). Степінь многочлена за таблицею значень**

Дано непорожній список цілих чисел `values` — значення многочлена в послідовних цілих точках: $P(0), P(1), \ldots, P(n-1)$, де `n = len(values)`. Потрібно визначити **найменший можливий степінь** многочлена, який набуває цих значень. Коефіцієнти шуканого многочлена не обов'язково цілі. Знаходити їх не потрібно.

Побудуйте таблицю різниць. Для цього від кожного наступного елемента списку відніміть попередній. Повторюйте цю дію для отриманих списків, доки не залишиться один елемент. Наприклад:

```text
Значення:          0  1  4  9  16
Перші різниці:       1  3  5  7
Другі різниці:         2  2  2
Треті різниці:          0  0
Четверті різниці:         0
```

Для многочлена другого степеня другі різниці сталі. Загалом шуканий степінь дорівнює кількості переходів до різниць, необхідних для отримання першого сталого рядка. **Сталим** називаємо непорожній рядок, усі елементи якого однакові; рядок з одним елементом також є сталим.

Реалізуйте чотири функції:

1. `differences(values)` повертає **новий список** різниць сусідніх елементів: елемент з індексом `i` дорівнює `values[i + 1] - values[i]`. Для списку з одним елементом поверніть `[]`.
2. `difference_table(values)` за допомогою послідовних викликів `differences` повертає список усіх рядків таблиці: від копії початкового списку до рядка з одним елементом включно. Таблицю потрібно будувати до кінця, навіть якщо сталий рядок зустрівся раніше. Кожен рядок має бути окремим списком.
3. `degree_from_table(table)` повертає індекс першого сталого рядка таблиці; нумерація починається з `0`. Гарантується, що `table` — повна таблиця, побудована за правилами попереднього пункту з непорожнього списку цілих чисел.
4. `polynomial_degree(values)` повертає результат за допомогою композиції:

```python
degree_from_table(difference_table(values))
```

**Домовленість:** для будь-якого сталого початкового списку, зокрема списку нулів, відповідь — `0`. Для нульового многочлена це прийнята в цій задачі умовність.

**Обмеження:** початковий список містить від 1 до 10 цілих чисел.
